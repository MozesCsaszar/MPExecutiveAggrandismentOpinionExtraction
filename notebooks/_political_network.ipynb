```python
import streamlit as st


def render_time_series_dashboard(
    df: pd.DataFrame,
    ts_metrics: pd.DataFrame,
    time_col: str = "time",
    group_col: str = "group",
    member_col: str | None = None,
    score_col: str = "score",
):
    st.header("Time-Series Opinion Analysis")

    tab1, tab2, tab3, tab4 = st.tabs(
        [
            "Trends",
            "Heatmaps",
            "Dynamics",
            "Individuals",
        ]
    )

    with tab1:
        st.subheader("Mean Opinion and Polarization")

        st.plotly_chart(
            plotly_mean_opinion_over_time(
                ts_metrics,
                time_col=time_col,
                group_col=group_col,
            ),
            use_container_width=True,
        )

        st.plotly_chart(
            plotly_polarization_over_time(
                ts_metrics,
                time_col=time_col,
                group_col=group_col,
            ),
            use_container_width=True,
        )

        st.plotly_chart(
            plot_divergence_from_population(
                ts_metrics,
                time_col=time_col,
                group_col=group_col,
            ),
            use_container_width=True,
        )

    with tab2:
        st.subheader("Heatmaps")

        st.plotly_chart(
            plot_opinion_heatmap(
                ts_metrics,
                time_col=time_col,
                group_col=group_col,
            ),
            use_container_width=True,
        )

        st.plotly_chart(
            plot_polarization_heatmap(
                ts_metrics,
                time_col=time_col,
                group_col=group_col,
            ),
            use_container_width=True,
        )

    with tab3:
        st.subheader("Dynamics")

        st.plotly_chart(
            plot_phase_diagram(
                ts_metrics,
                time_col=time_col,
                group_col=group_col,
            ),
            use_container_width=True,
        )

        st.plotly_chart(
            plot_opinion_velocity(
                ts_metrics,
                time_col=time_col,
                group_col=group_col,
            ),
            use_container_width=True,
        )

        st.plotly_chart(
            plot_opinion_acceleration(
                ts_metrics,
                time_col=time_col,
                group_col=group_col,
            ),
            use_container_width=True,
        )

        st.plotly_chart(
            plot_group_stability(
                ts_metrics,
                group_col=group_col,
            ),
            use_container_width=True,
        )

        st.plotly_chart(
            plot_pairwise_group_distances(
                ts_metrics,
                time_col=time_col,
                group_col=group_col,
            ),
            use_container_width=True,
        )

    with tab4:
        st.subheader("Individual Trajectories")

        if member_col is None or member_col not in df.columns:
            st.info("No member column provided.")
        else:
            st.plotly_chart(
                plot_individual_trajectories(
                    df,
                    time_col=time_col,
                    group_col=group_col,
                    member_col=member_col,
                    score_col=score_col,
                ),
                use_container_width=True,
            )
```

```python
def plot_opinion_heatmap(
    ts_metrics: pd.DataFrame,
    time_col: str = "time",
    group_col: str = "group",
):
    pivot = ts_metrics.pivot(index=group_col, columns=time_col, values="mean_opinion")

    fig = px.imshow(
        pivot,
        color_continuous_scale="RdBu",
        zmin=-1,
        zmax=1,
        aspect="auto",
        title="Mean Opinion Heatmap",
    )

    fig.update_layout(
        xaxis_title="Time", yaxis_title="Group", coloraxis_colorbar_title="Mean opinion"
    )

    return fig


def plot_polarization_heatmap(
    ts_metrics: pd.DataFrame,
    time_col: str = "time",
    group_col: str = "group",
):
    pivot = ts_metrics.pivot(index=group_col, columns=time_col, values="polarization")

    fig = px.imshow(
        pivot,
        color_continuous_scale="Viridis",
        zmin=0,
        zmax=2,
        aspect="auto",
        title="Polarization Heatmap",
    )

    fig.update_layout(
        xaxis_title="Time", yaxis_title="Group", coloraxis_colorbar_title="Polarization"
    )

    return fig
```
